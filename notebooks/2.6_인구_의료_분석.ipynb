```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\KDS-22\Desktop\Project\data\processed\02_의료기관1_v3.csv")

print(df.head())
```

```python
avg_med = df["의료기관수"].mean().round(2)
median_med = df["의료기관수"].median().round(2)

print(avg_med)
print(median_med)
```

```python
def classify_access(row):

    if row["의료기관수"] >= median_med:
        return "일반 의료기관 공급 양호"

    elif row["공공 기관 수"] > 0:
        return "일반 의료기관 부족·공공 보완"

    else:
        return "일반·공공 의료자원 모두 부족"

df["의료접근성유형"] = df.apply(classify_access, axis=1)

print(df.head(10))
```

```python
print(df["의료접근성유형"].value_counts())
```

```python
access_result = df[
    [
        "지역", "65세이상", "고령인구비율", "의료기관수", "공공 기관 수", "의료접근성유형"
    ]
].sort_values(["의료접근성유형", "65세이상"], ascending=[True, False])

print(access_result)
```

```python
df["고령인구1000명당공공기관수"] = (df["공공 기관 수"] / df["65세이상"] * 1000).round(2)

print(df.head())
```

```python
print(df[["지역", "고령인구1000명당의료기관수", "고령인구1000명당공공기관수"]].sort_values("고령인구1000명당의료기관수"))
```

```python
fig, (ax1, ax2) = plt.subplots
```

```python
median_density = df["고령인구1000명당의료기관수"].median()
avg_old = df["65세이상"].mean()

print(median_density)
print(avg_old)
```

```python
df["공공기관여부"] = df["공공 기관 수"] > 0
```

```python
def classify_access(row):

    if row["고령인구1000명당의료기관수"] >= median_density:
        return "의료 접근 양호"

    elif row["공공 기관 수"] > 0:
        return "공공 보완 지역"

    else:
        return "우선 검토 지역"


df["의료접근성유형"] = df.apply(classify_access, axis=1)
```

```python
print(df[["지역", "고령인구1000명당의료기관수", "공공 기관 수", "의료접근성유형"]].sort_values("고령인구1000명당의료기관수"))
```

```python
def classify_region(row):

    if row["고령인구1000명당의료기관수"] >= median_density:
        return "의료 접근 양호"

    elif row["공공 기관 수"] > 0:
        return "공공기관 거점 활용 후보"

    elif row["65세이상"] >= avg_old:
        return "방문의료 우선 검토 후보"

    else:
        return "일반 관리 지역"

df["지역유형"] = df.apply(classify_region,axis=1)
```

```python
result = df[
    [
    "지역", "65세이상", "고령인구1000명당의료기관수", "공공 기관 수", "지역유형"
    ]
].sort_values(["지역유형", "65세이상"], ascending=[True, False])

print(result)
```
