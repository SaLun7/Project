```python
print(df_region[["지역", "65세이상", "고령인구비율"]].head(10))
print(df_region.shape)
```

```python
import matplotlib.pyplot as plt
import matplotlib

matplotlib.rc("font", family="Malgun Gothic")
plt.rcParams["axes.unicode_minus"] = False

avg_rate = df_region["고령인구비율"].mean()
avg_old = df_region["65세이상"].mean()

plt.figure(figsize=(10, 7))
plt.scatter(
    df_region["고령인구비율"],
    df_region["65세이상"]
)

for _, row in df_region.iterrows():
    plt.text(
        row["고령인구비율"] + 0.3,
        row["65세이상"] + 50,
        row["지역"],
        fontsize = 8
    )
    
plt.title("천안시 고령인구 비율과 65세 이상 인구의 관계")
plt.xlabel("고령인구 비율(%)")
plt.ylabel("65세 이상 인구 수")
plt.grid(alpha=0.3)

plt.axvline(avg_rate, color="red", linestyle="--")
plt.axhline(avg_old, color="blue", linestyle="--")

plt.savefig("고령인구 비율과 65세이상 인구의 관계.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
def classify_region(row):
    if row["고령인구비율"] >= avg_rate and row["65세이상"] >= avg_old:
        return "고비율·대규모"
    elif row["고령인구비율"] >= avg_rate:
        return "고비율·소규모"
    elif row["65세이상"] >= avg_old:
        return "저비율·대규모"
    else:
        return "저비율·소규모"

df_region["인구유형"] = df_region.apply(classify_region, axis=1)

print(df_region[["지역", "65세이상", "고령인구비율", "인구유형"]].sort_values(["인구유형", "고령인구비율"], ascending=[True, False]))
```

```python
plt.figure(figsize=(10, 7))
plt.scatter(
    df_region["고령인구비율"],
    df_region["65세이상"],
    
)

highlight = ["성환읍", "목천읍", "청룡동", "동면"]

for _, row in df_region.iterrows():
    if row["지역"] in highlight:
        plt.scatter(
            row["고령인구비율"],
            row["65세이상"],
            color="red",
            s=100
        )
        
        plt.text(
            row["고령인구비율"] + 0.5,
            row["65세이상"] + 50,
            row["지역"],
            fontsize=9
        )

    else:
        plt.scatter(
            row["고령인구비율"],
            row["65세이상"],
            color="gray",
            s=40
        )


        plt.text(
            avg_rate + 1,
            df_region["65세이상"].max() * 1,
            "고비율·대규모",
            fontsize=11
        )

        plt.text(
            avg_rate + 1,
            df_region["65세이상"].min(),
            "고비율·소규모",
            fontsize=11
        )

        plt.text(
            df_region["고령인구비율"].min(),
            df_region["65세이상"].max() * 1,
            "저비율·대규모",
            fontsize=11
        )

        plt.text(
            df_region["고령인구비율"].min(),
            df_region["65세이상"].min(),
            "저비율·소규모",
            fontsize=11
        )


    
plt.title("천안시 고령인구 비율과 65세 이상 인구의 관계")
plt.xlabel("고령인구 비율(%)")
plt.ylabel("65세 이상 인구 수")
plt.grid(alpha=0.3)

plt.axvline(
    avg_rate,
    color="red",
    linestyle="--",
    label="평균 고령인구비율"
)

plt.axhline(
    avg_old,
    color="blue",
    linestyle="--",
    label="평균 고령인구수"
)

plt.legend()

plt.savefig("고령인구 비율과 65세이상 인구의 관계.png", dpi=300, bbox_inches="tight")

plt.show()
```
